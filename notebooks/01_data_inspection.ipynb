```python
def missing_value_summary(dataframe):
    missing_count = dataframe.isnull().sum()
    missing_percentage = (missing_count / len(dataframe)) * 100

    summary = pd.DataFrame({
        "missing_count": missing_count,
        "missing_percentage": missing_percentage
    })

    return summary[summary["missing_count"] > 0]
```

```python
print("ACCOUNTS")
display(missing_value_summary(accounts))

print("\nSUBSCRIPTIONS")
display(missing_value_summary(subscriptions))

print("\nFEATURE USAGE")
display(missing_value_summary(feature_usage))

print("\nSUPPORT TICKETS")
display(missing_value_summary(support_tickets))

print("\nCHURN EVENTS")
display(missing_value_summary(churn_events))
```
